3. Постройте графики зависимости пути от времени для тел, одно из ко­торых движется с постоянной скоростью $v_1 = 7 \frac{m}{s}$, а второе $v_2 = 3 \frac{m}{s}$

````python
import matplotlib.pyplot as plt
import numpy as np

plt.figure(figsize=(5, 5))

scale=10
v1=np.r_[0, 7]*scale
v2=np.r_[0, 3]*scale
timeline = np.r_[0, 1]*scale

plt.plot(timeline, v1, "r-", label=r"$v_1 = 7\,\frac{\text{м}}{\text{с}}$")
plt.plot(timeline, v2, "g-", label=r"$v_2 = 3\,\frac{\text{м}}{\text{с}}$")

plt.xlabel(r"$t$, с")
plt.ylabel(r"$s$, м")
plt.legend(fontsize=11)
plt.grid(True, linestyle="--", alpha=0.5)

plt.show()
````

7. Первую половину пути из Москвы в Подольск автомобиль ехал со скоростью v1 = 90 км/ч. Оставшийся путь s = 20 км из-за поломки при­ v2 = 10 км/ч. 

- Сколько времени двигался ав­томобиль до Подольска? 
- С какой постоянной скоростью должен дви­гаться велосипедист, чтобы преодолеть это расстояние за то же вре­мя, что и автомобиль?

````python
from fractions import Fraction

v1=90
s1=20
s2=20
v2=10

t1=Fraction(s1, v1)
t2=s2/v2

print("Ответы:")
print(f"- Автомобиль двигался {t2} часа {(60*60*t1)//60} минут {(60*60*t1)%60} секунд")
print(f"- Чтобы преодолеть {s1+s2} км. велосипедист должен двигаться со скоростью {(s1+s2)/(t1+t2):.0f} км/ч")
````
