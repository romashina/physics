**Ответ**: Если велосипедист едет, то он не движется относительно седла. Если велосипед поставить на стоянку, то велосипедист будет двигаться относительно седла.

3. Возраст человека 12 лет. Каков возраст этого человека в единицах СИ? Напомним, что каждый четвёртый год — високосный.

```python
age = 12
print("Ответ:", (age*365+12/4)*24*60*60, "c")
```

4. Петром I вместо локтя в обиход было введено понятие «аршин» (шаг). 1 аршин = 28 дюймов = 16 вершков = 71,12 см. Выразите: 

- а) 30 дюй­мов в метрах; 
- б) 2 аршина 16 дюймов 3 вершка в санти­метрах.

```python
arshin = 71.12 
inch = arshin/28
vershok = arshin/16 

print("Ответы:")
print("а) ", 30*inch/100, "м")
print("б) ", 2*arshin+16*inch+3*vershok, "см")
```
