4. Петром I вместо локтя в обиход было введено понятие «аршин» (шаг). 1 аршин = 28 дюймов = 16 вершков = 71,12 см. Выразите: 

- а) 30 дюй­мов в метрах; 
- б) 2 аршина 16 дюймов 3 вершка в санти­метрах.

````python
arshin = 71.12 
inch = arshin/28
vershok = arshin/16 

print("Ответы:")
print("а) ", 30*inch/100, "м")
print("б) ", 2*arshin+16*inch+3*vershok, "см")
````

8. Автобус проходит расстояние 30 км между двумя посёлками со ско­ростью 60 км/ч. В обратном направлении этот путь он проходит со скоростью 50 км/ч. Найдите среднюю скорость автобуса за всё время движения.

````python
s=30
v1=60
v2=50
t1=s/v1
t2=s/v2
print("Ответ:", ((2*s)/(t1+t2)), "км/ч")
````
