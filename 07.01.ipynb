## Упражнение 3

1. Чтобы огурцы продолжительное время оставались малосольными, рассол с огурцами необходимо хранить в холодном помещении. По­чему?

**Ответ**: При понижении температуры уменьшается скорость хаотического (теплового) движения молекул воды и ионов соли. В результате процесс диффузии (проникновения соли в ткани огурца) резко замедляется, и огурцы дольше сохраняют состояние малосольных.

2. Если поместить на дно стакана крупинку краски, а затем осторожно налить в стакан воду, то через некоторое время вода приобретает окраску, хотя стакан стоял неподвижно. Объясните явление.

**Ответ**: Наблюдаемое явление называется диффузией. Вследствие непрерывного хаотического теплового движения молекулы краски и воды самопроизвольно проникают в промежутки между молекулами друг друга, что со временем приводит к их полному перемешиванию и равномерному окрашиванию всей жидкости.

3. Почему цветные пятна на столе или полу значительно легче удалить сразу после того, как они туда попали, чем спустя некоторое время?

**Ответ**: С течением времени происходит диффузия — молекулы красящего вещества проникают глубоко в поры и межмолекулярные промежутки материала пола или стола. Кроме того, растворитель испаряется, и краситель закрепляется в толще материала, из-за чего извлечь его оттуда становится значительно сложнее, чем смахнуть с поверхности в первые секунды.

## Упражнение 4

1. Кузнец раскалил два куска железа, наложил их друг на друга и уда­рил молотом. Почему в результате этого куски соединились?

**Ответ**: При кузнечной сварке срабатывают три связанных фактора:

1. Пластическая деформация: Раскалённое железо становится пластичным. Удар молота сглаживает микронеровности и разрушает оксидную плёнку, сближая чистые поверхности металла на расстояние порядка размера атома, где начинают действовать силы межатомного (металлического) притяжения.
2. Ускорение диффузии: Высокая температура передаёт атомам кинетическую энергию, благодаря чему они начинают активно перескакивать через границу раздела и проникать в кристаллическую решётку соседнего куска.
3. Давление (удар): Механическое давление обеспечивает плотный физический контакт по всей площади соединения, позволяющий диффузии и межатомным связям сработать мгновенно.

2. Клей позволяет прочно соединять тела. Объясните, почему это про­исходит.

**Ответ**:  Жидкий клей заполняет микронеровности склеиваемых поверхностей и сближает свои молекулы с молекулами деталей на расстояние порядка их размеров ($\approx 10^{-9}\text{ м}$). На таком расстоянии возникают прочные силы межмолекулярного притяжения (адгезия). При высыхании или застывании клей твердеет, образуя прочные связи и внутри самого себя (когезия), что создаёт монолитное соединение.

## Упражнение 5

1. На чувствительных весах уравновесили два одинаковых стакана. Стаканы заполнили до краёв водой: один — горячей, а другой — хо­лодной. Равновесие весов нарушилось. Объясните почему.

**Ответ**: Равновесие нарушилось: стакан с холодной водой перевесил горячий.
При нагревании вода расширяется, поэтому плотность горячей воды меньше плотности холодной. Поскольку объёмы обоих стаканов одинаковы, масса горячей воды в стакане окажется меньше массы холодной воды ($m = \rho \cdot V$). Дополнительно разницу в массе увеличивает интенсивное испарение горячей воды.

2. Что будет происходить с газом, если его постоянно охлаждать? Объ­ясните своё предположение с позиций молекулярного строения ве­щества.

**Ответ**: При постоянном охлаждении газ сначала сконденсируется в жидкость, а при дальнейшем понижении температуры отвердеет и станет твёрдым телом (кристаллом).
Температура характеризует скорость движения частиц. При охлаждении скорость и кинетическая энергия молекул газа уменьшаются, из-за чего силы межмолекулярного притяжения начинают преобладать над тепловым движением. Расстояние между молекулами уменьшается до размеров самих молекул, и газ превращается в жидкость. При продолжающемся охлаждении частицы теряют возможность перемещаться и застывают в фиксированных узлах кристаллической решётки, образуя твёрдое тело.
