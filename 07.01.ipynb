## 5. Точность и погрешность измерений

### Вопрос

1. За счёт чего возникает погрешность при измерении? При измерении и из-за прибора
2. Что называется абсолютной погрешностью измерения? Отклонение измерения от истинного значения.
3. Из чего складывается абсолютная погрешность измерения? Сумма *погрешности отсчета* и *инструментальной погрешности*
4. Как записать физическую величину с учётом абсолютной погрешности измерения? А = a ± Δa
