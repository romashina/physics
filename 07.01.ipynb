1. Капля масла объёмом $0,003\ мм^3$ растеклась по поверхности воды тонким слоем и заняла площадь $300\ см^2$. 
Принимая толщину слоя равной диаметру молекулы масла, определите этот диаметр.

```python
v=0.003/(1000*1000*1000)
s=300/(100*100)
v/s
```

**Ответ**: Диаметр молекулы $10^{-10}\ м$

2. Вы не раз наблюдали, как капли масляной жидкости растекаются по поверхности воды, образуя тонкие плёнки. 
Почему эти плёнки (при небольшом количестве масла) не могут покрыть всю поверхность во­доёма?

**Ответ**: Потому что есть конечная тонкость пленки, выраженная в одной частице, тоньше одной молекулы быть не может.

3. Мощным прессом удаётся сжать даже такой плотный металл, как свинец, до 0,85 начального объёма. Чем объяснить такое значитель­ное сжатие свинца?

**Ответ**: наличием пространства между атомами.

4. Число молекул воздуха в ограниченном объёме в $1\ см^3$ очень вели­ко — $2,7 \cdot 10^{19}$. 
Если сделать щель, через которую будет просачивать­ся по миллиону молекул в секунду, за сколько лет все молекулы по­кинут этот объём?

```python
molecules = 2.7 * 10 ** 19
one_second = 1000_000
total_seconds = molecules / one_second
year =total_seconds/(60*60*24*365)

print(f"{year=}")
```

**Ответ**: $\approx 856164.38\ лет$
